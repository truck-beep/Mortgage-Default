```python
import pandas as pd
import numpy as np
import pandas_datareader.data as pdr
import datetime
from sklearn.linear_model import LinearRegression
import sqlite3
```

```python
np.random.seed(111)

def avoid_mixed_types(df):
    object_cols = df.select_dtypes(include=["object"]).columns
    df[object_cols] = df[object_cols].astype("string")
    return

# Only take a sample of all mortgage observations
def sample_mortgages(orig_keep_cols,perf_keep_cols,sample_size):

    # mortgage files have headers in a separate file
    orig_headers_file = pd.read_excel("raw data/file_layout_july_2026.xlsx",sheet_name="Origination Data File",header=1)
    orig_headers = orig_headers_file.iloc[:,1]
    perf_headers_file = pd.read_excel("raw data/file_layout_july_2026.xlsx",sheet_name="Monthly Performance Data File",header=1)
    perf_headers = perf_headers_file.iloc[:,1]

    # initialize the dataframe
    df = pd.DataFrame()


    for i in range(1999,2027):
        print(f"Year {i}")
        print(f"Origination")
        # Make sure Postal Code column isn't converted to numeric and loses its leading zeroes
        orig_data = pd.read_csv(f"raw data/sample_orig_{i}.txt",sep="|",header=None,dtype={18:"string"})
        orig_data.rename(columns=orig_headers,inplace=True)
        orig_data = orig_data.loc[:,orig_keep_cols]

        print(f"Performance")
        # Make sure Loan Delinquency or Zero Balance Code columns don't lose leading zeroes
        perf_data = pd.read_csv(f"raw data/sample_perf_{i}.txt",sep="|",header=None,dtype={3:"string",8:"string"})
        perf_data.rename(columns=perf_headers,inplace=True)
        perf_data = perf_data.loc[:,perf_keep_cols]

        orig_data = orig_data.sample(n=sample_size)

        # We add the year's observations, using both loan origination and performance informations
        df = pd.concat([df,orig_data.merge(perf_data,how="inner",on="Loan Identifier")])
        # Only same some of the loans

    # Avoid a situation down the line where a type-specific function fails to work
    avoid_mixed_types(df)

    return df

# Note that adjustable-rate mortgages are no longer offered by Freddie Mac and Fannie Mae
orig_cols = ["Loan Identifier","Original UPB","Postal Code","Classic FICO®","Occupancy Status","Original Interest Rate",
                  "Original Combined Loan-to-Value (CLTV)","Original Debt-to-Income (DTI) Ratio","Original Loan-to-Value (LTV)",
                  "First Time Homebuyer Indicator","HARP Indicator","First Payment Date"]
perf_cols = ["Loan Identifier","Period","Current Actual UPB","Current Loan Delinquency Status","Loan Age","Modification Flag",
                  "Zero Balance Code","Payment Deferral Flag","Estimated Loan-to-Value (ELTV)",
                  "Delinquency Due to Disaster","Current Period Modification Costs","Borrower Assistance Plan"]

# Can do sample size of 2000
data = sample_mortgages(orig_cols,perf_cols,2000)

data["Period"] = pd.to_datetime(data["Period"],format="%Y%m")
data["First Payment Date"] = pd.to_datetime(data["First Payment Date"],format="%Y%m")

```

```python
# Reasons to default: cant pay anymore or property not worth the mortgage
# For the former, it is based off sudden changes in situation (i.e. unemployment, disaster)
# For the latter, it is based off of whether estimated LTV > 100 percent (which aren't tied to a relief program)

# Freddie Mac uses various values to denote missing information
# First I will convert them all to nan

# Loan-related columns
data.loc[data["Classic FICO®"]==9999,"Classic FICO®"] = pd.NA
data.loc[data["First Time Homebuyer Indicator"]=="9","First Time Homebuyer Indicator"] = pd.NA
data.loc[data["Occupancy Status"]=="9","Occupancy Status"] = pd.NA
data.loc[data["Original Combined Loan-to-Value (CLTV)"]==999,"Original Combined Loan-to-Value (CLTV)"] = pd.NA
data.loc[data["Original Debt-to-Income (DTI) Ratio"]==999,"Original Debt-to-Income (DTI) Ratio"] = pd.NA
data.loc[data["Original Loan-to-Value (LTV)"]==999,"Original Loan-to-Value (LTV)"] = pd.NA
data.loc[data["Postal Code"]=="000","Postal Code"] = pd.NA
# Monthly columns
data.loc[data["Estimated Loan-to-Value (ELTV)"]==999,"Estimated Loan-to-Value (ELTV)"] = pd.NA
```

```python
# We need to decide how to deal with missing values without data leakage


# Some loan-related columns have very few missing values so we exclude them,
# so we exclude loans that have missing values for those
data.dropna(subset=["Postal Code","Original Loan-to-Value (LTV)","Occupancy Status","Original Combined Loan-to-Value (CLTV)","Classic FICO®"],
            inplace=True)

# Certain columns represent indicators where they are explicitly defined in the user guide as representing "No" or 0:
data.loc[data["First Time Homebuyer Indicator"].isna(),"First Time Homebuyer Indicator"] = "No"
data.loc[data["Borrower Assistance Plan"].isna(),"Borrower Assistance Plan"] = "No"
data.loc[data["Delinquency Due to Disaster"].isna(),"Delinquency Due to Disaster"] = "No"
data.loc[data["Modification Flag"].isna(),"Modification Flag"] = "No"
data.loc[data["Payment Deferral Flag"].isna(),"Payment Deferral Flag"] = "No"
data.loc[data["Current Period Modification Costs"].isna(),"Current Period Modification Costs"] = 0
data.loc[data["Zero Balance Code"].isna(),"Zero Balance Code"] ="No"

# Certain columns have extreme values labelled as missing which may bring statistical bias concerns
# For Estimated Loan-to-Value (ELTV), we will estimate missing values using regional Housing Price Indices

# For Original Debt-to-Income (DTI) Ratio, values above 65% are reported as missing 
# The HARP program always has DTI missing and accounts for most missing DTI values
# HARP (2009-2018) was a refinancing program for those with property worth less than mortgage after 2008 crisis
# We remove all loans with missing values not part of the HARP program
missing_DTI_not_HARP = data["Original Debt-to-Income (DTI) Ratio"].isna() & (data["HARP Indicator"] == "N")
data = data.loc[~missing_DTI_not_HARP,:]


# Then using a training set of loans originated up to 2008, we impute missing values part of the HARP program through linear regression

# Note that since we only want one observation per loan since we are predicting a loan origination field
# so using time-varying loan characteristics doesn't make sense
DTI_train = (data.loc[data["First Payment Date"] < "2009-01-01",
                        ["Loan Identifier","Classic FICO®", "Original Loan-to-Value (LTV)","Original Debt-to-Income (DTI) Ratio"]]
                        .drop_duplicates())

DTI_test = (data.loc[data["Original Debt-to-Income (DTI) Ratio"].isna() & (data["First Payment Date"] >= "2009-01-01"),
                        ["Loan Identifier","Classic FICO®", "Original Loan-to-Value (LTV)","Original Debt-to-Income (DTI) Ratio"]]
                        .drop_duplicates())

reg = LinearRegression().fit(DTI_train[["Classic FICO®", "Original Loan-to-Value (LTV)"]], DTI_train["Original Debt-to-Income (DTI) Ratio"])
# Signs of coefficients make sense
print(reg.coef_)
DTI_test.loc[:,"Original Debt-to-Income (DTI) Ratio"] = reg.predict(DTI_test[["Classic FICO®", "Original Loan-to-Value (LTV)"]])

DTI_map = DTI_test.set_index("Loan Identifier")["Original Debt-to-Income (DTI) Ratio"]
data["Original Debt-to-Income (DTI) Ratio"] = (data["Original Debt-to-Income (DTI) Ratio"]
                                                    .fillna(data["Loan Identifier"].map(DTI_map)))

del DTI_map
del DTI_test
del DTI_train
```

```python
# We compute the default variable which we define as not being able to continue making payments without intervention in the next 12 months
def default_condition(row):
    return ((row["Current Loan Delinquency Status"] not in ["00","01","02","XX"])
        or  (row["Zero Balance Code"] in ["03","09"])
        or  (row["Delinquency Due to Disaster"] == "Y")
        or  (row["Modification Flag"] == "Y")
        or  (row["Current Period Modification Costs"] > 0)
        or  (row["Payment Deferral Flag"] == "C")
        or  (row["Borrower Assistance Plan"] in ["F","T","R"]))

data["Default"] = data.apply(default_condition,axis=1)

# Once default happens, I need to delete all observations following that default with that loan identifier
# Hence, I need the first occurence of monthly default
first_default = data.groupby("Loan Identifier").apply(lambda x: x.loc[x["Default"] == True,"Period"].min(),include_groups=False)
data["First Default Date"] = data["Loan Identifier"].map(first_default)
# Note that we also want to delete the observation when the default happens
data = data.loc[(data["Period"] < data["First Default Date"]) | (data["First Default Date"].isna()),:]

# Then we construct the 12 month default variable as whether there is a monthly default in the next 12 months
# This does bring concerns of serial correlation which has problems for standard errors but not for predictions
data["Default Next 12"] = data.apply(lambda x: x["Period"] >= (x["First Default Date"] - pd.tseries.offsets.DateOffset(months=12)),axis=1)

data.drop(labels=["Current Loan Delinquency Status","Zero Balance Code","Delinquency Due to Disaster","Modification Flag",
                       "Current Period Modification Costs","Payment Deferral Flag","Borrower Assistance Plan","Default",
                       "First Default Date"],axis=1,inplace=True)
```

```python
# Estimated LTV is an important measure of whether the borrower has an incentive to voluntarily default
# However, this field is only populated from March 2015
# We use the Housing Price Index (hpi) dataset to construct our own measure

def prep_hpi():
    df = pd.DataFrame()
    # Observations are quarterly so to make it monthly we would need 3 times the observations
    for i in range(3):
        hpi = pd.read_excel("raw data/hpi_at_3zip.xlsx",skiprows=4,dtype={"Three-Digit ZIP Code":"string"})
        hpi["Three-Digit ZIP Code"] = hpi["Three-Digit ZIP Code"].str.zfill(3)
        avoid_mixed_types(hpi)
        
        hpi.sort_values(by=["Three-Digit ZIP Code","Year","Quarter"],inplace=True)
        hpi["Index (NSA) Lag"] = hpi.groupby("Three-Digit ZIP Code")["Index (NSA)"].shift()
        hpi["Quarter Period"] = i+1
        df = pd.concat([df,hpi])

    # Here we use linear interpolation to convert quarterly indices to monthly indices
    df["Housing Index"] = (df["Index (NSA)"] - df["Index (NSA) Lag"]) * (df["Quarter Period"]/3) + df["Index (NSA) Lag"]

    df["date"] = df.apply(lambda x: f"01/{(x["Quarter"] - 1) * 3 + x["Quarter Period"]}/{x["Year"]}",axis=1)
    df["date"] = pd.to_datetime(df["date"],format="%d/%m/%Y")

    df.rename(mapper={"Three-Digit ZIP Code":"Postal Code","date":"Period"},axis=1,inplace=True)
    df = df[["Period","Postal Code","Housing Index"]]
    df.dropna(inplace=True)

    return df

# We need to determine the change in housing price, so we need index at loan origination and at current period
hpi = prep_hpi()

data = pd.merge(data,hpi,how="left",on=["Period","Postal Code"])
hpi.rename(mapper={"Period":"First Payment Date","Housing Index":"Origination Housing Index"},axis=1,inplace=True)
data = pd.merge(data,hpi,how="left",on=["First Payment Date","Postal Code"])

data["Computed Property Price"] = (100 * (data["Original UPB"] / data["Original Loan-to-Value (LTV)"]) * 
                                   (data["Housing Index"] / data["Origination Housing Index"]))
data["Computed LTV"] = (data["Current Actual UPB"] / data["Computed Property Price"]) * 100

# Estimates are very accurate with a few extreme cases
# Hence, we replace missing values of Estimated LTV with our computed version
print(data[["Estimated Loan-to-Value (ELTV)","Computed LTV"]])
print()
print(data[["Estimated Loan-to-Value (ELTV)","Computed LTV"]].corr())
print()
print((data["Computed LTV"] - data["Estimated Loan-to-Value (ELTV)"]).describe(percentiles=[0.1,0.25,0.5,0.75,0.9]))

data["Estimated Loan-to-Value (ELTV)"] = data["Estimated Loan-to-Value (ELTV)"].fillna(data["Computed LTV"])

# Note that the HPI dataset does not have indices for 006-009, 102, and 969 which represent Puerto Rico, part of New York, and Guam respectively
# As it represents a very small proportion of mortgage loans, we remove those observations
data = data[~data["Postal Code"].isin(["006", "007", "008", "009", "102", "969"])]


data.drop(labels=["Computed Property Price", "Computed LTV", "Original UPB", "Current Actual UPB",
                       "Housing Index", "Origination Housing Index", "First Payment Date"],axis=1,inplace=True)

del hpi
```

```python
# Here we construct our macroeconomic variable database and then add it to our mortgage database

start_date = datetime.datetime(1998, 1, 1)
end_date = datetime.datetime(2026, 3, 1)

# Variables: real GDP proxy (GDP is quarterly otherwise), CPI, real wages, Credit card borrowing rate (fed funds rate),
# car loans rate (5-year rate), unemployment rate
# All in monthly terms
tickers = ["INDPRO","CPIAUCSL","CES0500000003","FEDFUNDS","GS5","UNRATE"]
macro = pdr.DataReader(tickers, 'fred', start_date, end_date)
avoid_mixed_types(macro)

macro.reset_index(inplace=True)
macro.rename(mapper={"DATE":"Period"},axis=1,inplace=True)
macro["Period"] = pd.to_datetime(macro["Period"],format="%d-%m-%Y")

# There is surprisingly 1 missing observation in October 2025 for UNRATE and CPIAUCSL, so we interpolate it
macro["CPIAUCSL"] = macro["CPIAUCSL"].interpolate(method="linear")
macro["UNRATE"] = macro["UNRATE"].interpolate(method="linear")


# Our wages variable is not inflation-adjusted so it is adjusted
macro["Real Wage"] = (macro["CES0500000003"] / macro["CPIAUCSL"]) * 100

# For some variables, we convert them to monthly percentage changes as we do not care about the absolute value
macro["INDPRO change"] = macro["INDPRO"].pct_change() * 100
macro["CPI change"] = macro["CPIAUCSL"].pct_change() * 100
macro["Real Wage change"] = macro["Real Wage"].pct_change() * 100

macro.drop(labels=["Real Wage","INDPRO","CPIAUCSL","CES0500000003"],inplace=True,axis=1)


# Since we are looking at probability of default in next 12 months, and there tends to be a delay between event and default
# due to 90 days delinquency criteria, exhausting credit cards, receiving unemployment benefits, bureaucracy for aid programs, etc.,
# we need to find how far lagged do we want our variables

# For macro variables, we want periods of larger than one month since we care about sustained shocks, but we don't want too large periods
# otherwise shocks may be averaged out. So we use 3-,6-,9-month averages.
macro_3 = macro.loc[:,macro.columns != "Period"].rolling(window=3).mean().add_suffix(" 3",axis=1)
macro_6 = macro.loc[:,macro.columns != "Period"].rolling(window=6).mean().add_suffix(" 6",axis=1)
macro_9 = macro.loc[:,macro.columns != "Period"].rolling(window=9).mean().add_suffix(" 9",axis=1)

macro = macro.loc[:,"Period"]
macro = pd.concat([macro,macro_3,macro_6,macro_9],axis=1)

# Note that Real Wage variables are the only ones with missing values
data = pd.merge(data,macro,how="left",on=["Period"])

del macro_3
del macro_6
del macro_9
del macro
```

```python
# Now we add weather disaster events since that directly affects the value of a house, and repairs may put them into delinquency
housing_assistance = pd.read_csv("raw data/HousingAssistanceOwners.csv",dtype={"zipCode":"string"}).loc[:,["disasterNumber","zipCode",
                "totalInspected","femaInspectedDamageBetween1And10000","femaInspectedDamageBetween10001And20000",
                "femaInspectedDamageBetween20001And30000","femaInspectedDamageGreaterThan30000","approvedForFemaAssistance",
                "repairReplaceAmount","totalDamage"]]
avoid_mixed_types(housing_assistance)
# Since FEMA disaster data has information up to 5-digit zipcodes, while our mortgage data has only 3-digit zipcodes, 
# we sum up all 5-digit zipcode information before analyzing them

housing_assistance["Postal Code"] = housing_assistance["zipCode"].apply(lambda x: x[0:3])

housing_assistance.drop(labels=["zipCode"],axis=1,inplace=True)

housing_assistance = housing_assistance.groupby(["disasterNumber","Postal Code"]).sum().reset_index()

# To avoid division by zero or small-sample variance, we further remove some observations
housing_assistance = housing_assistance.loc[(housing_assistance["totalInspected"] >= 30) &
                                            (housing_assistance["approvedForFemaAssistance"] >= 30),:]


housing_assistance["femaInspectedDamageGreaterThan0"] = (housing_assistance["femaInspectedDamageBetween1And10000"] + 
        housing_assistance["femaInspectedDamageBetween10001And20000"] + housing_assistance["femaInspectedDamageBetween20001And30000"] +
        housing_assistance["femaInspectedDamageGreaterThan30000"])



# Extreme disasters are determined by looking at 3 extreme disasters pre-2012, computing the proportion of inspections resulting
# in more than $30000 worth of damage aggregated across all zipcodes, and taking the smallest of the 3 numbers as the threshold
# Hurricane Katrina, Hurricane Ike, 2011 Super Tornado
# Thresholds are applied at the disaster, 3-digit zipcode level

examples_allowed = ((housing_assistance["disasterNumber"] == 1603) | (housing_assistance["disasterNumber"] == 1791) | 
                    (housing_assistance["disasterNumber"] == 1971))
examples = housing_assistance.loc[examples_allowed,:]
examples = examples.groupby("disasterNumber").sum().reset_index()

# We find a threshold of 3.4% rounded
print("Extreme disaster threshold")
print((examples["femaInspectedDamageGreaterThan30000"] / examples["totalInspected"]).min())

extreme_disaster = (housing_assistance["femaInspectedDamageGreaterThan30000"] / housing_assistance["totalInspected"]) >= 0.034
housing_assistance["Extreme Disaster"] = extreme_disaster

# We define moderate disaster events where at least 20 percent of house inspections resulted in damage > 0, 
# following https://www.tandfonline.com/doi/full/10.1080/10527001.2020.1838172#d1e376, that aren't extreme disasters
# Notably, nearly all observations are classified as moderate, meaning requiring a minimum number of 30 inspections/grants seems
# to correspond to the same requirement
moderate_disaster = (housing_assistance["femaInspectedDamageGreaterThan0"] / housing_assistance["totalInspected"]) >= 0.2
housing_assistance["Moderate Disaster"] = (moderate_disaster & ~extreme_disaster)
housing_assistance = housing_assistance.loc[(moderate_disaster | extreme_disaster),:]



# For each designated disaster, we compare average inspected damage to average payout as measure of relief assistance
# Note that total inspections includes those with 0 damage inspected, which we must exclude since they will not receive a payout
housing_assistance["average inspected damage"] = housing_assistance["totalDamage"] / housing_assistance["femaInspectedDamageGreaterThan0"]
housing_assistance["average payout"] = housing_assistance["repairReplaceAmount"] / housing_assistance["approvedForFemaAssistance"]


# We end up with cases where payout is greater than damage
# Since there is an asymmetric impact on mortgage default where payout greater than damage won't reduce mortgage default likelihood,
# we clip the values between 0 and 1
housing_assistance["Uncovered Damage Ratio"] = 1 - (housing_assistance["average payout"] / housing_assistance["average inspected damage"])
housing_assistance["Uncovered Damage Ratio"] = housing_assistance["Uncovered Damage Ratio"].clip(lower=0, upper=1)



housing_assistance = housing_assistance.loc[:,["disasterNumber","Postal Code","Extreme Disaster",
                                               "Moderate Disaster","Uncovered Damage Ratio"]]



# We still need to determine which months a disaster takes place
disasters = (pd.read_csv("raw data/DisasterDeclarationsSummaries.csv")
                        .loc[:,["disasterNumber","incidentBeginDate","incidentEndDate"]]
                        .drop_duplicates())
avoid_mixed_types(disasters)

disasters["incidentEndDate"].fillna(disasters["incidentBeginDate"],inplace=True)

disasters["incidentBeginDate"] = disasters["incidentBeginDate"].apply(lambda x: x[0:10])
disasters["incidentBeginDate"] = pd.to_datetime(disasters["incidentBeginDate"])
disasters["incidentEndDate"] = disasters["incidentEndDate"].apply(lambda x: x[0:10])
disasters["incidentEndDate"] = pd.to_datetime(disasters["incidentEndDate"])

disaster_times = []

for _, row in disasters.iterrows():
        month_range = pd.date_range(start=row["incidentBeginDate"].replace(day=1), end=row["incidentEndDate"].replace(day=1), freq="MS")

        # We create a new row for each month a disaster takes place during
        for date in month_range:
                disaster_times.append({"disasterNumber": int(row["disasterNumber"]),"Period": date})

disaster_times = pd.DataFrame(disaster_times).drop_duplicates()
disaster_times["Period"] = pd.to_datetime(disaster_times["Period"])


housing_assistance = pd.merge(housing_assistance,disaster_times,on=["disasterNumber"],how="left")
housing_assistance.drop(labels=["disasterNumber"],axis=1,inplace=True)

# We now need to create only one observation at most per zipcode and month
# We make extreme disasters have priority over moderate disasters, and choose the higher uncovered damage ratio

housing_assistance = housing_assistance.groupby(["Period","Postal Code"]).agg({"Moderate Disaster":"max",
                                                "Extreme Disaster":"max","Uncovered Damage Ratio":"max"}).reset_index()

housing_assistance.loc[(housing_assistance["Moderate Disaster"]==True) & (housing_assistance["Extreme Disaster"]),"Moderate Disaster"] = False

# Since there may be a delay between disaster and inability to pay debts, we check if disaster happened in last 3-,6-,9- months.



def disaster_lag(data_to_lag,lag):
        df_0 = data_to_lag.copy()
        df_0 = df_0.rename(columns={"Moderate Disaster":"Moderate Disaster Lag 0", "Extreme Disaster":"Extreme Disaster Lag 0"})
        moderate_cols = ["Moderate Disaster Lag 0"]
        extreme_cols = ["Extreme Disaster Lag 0"]
        for i in range(1,lag):
                df_lag = data_to_lag.copy()
                df_lag["Period"] = df_lag["Period"] + pd.tseries.offsets.DateOffset(months=i)
                df_lag = df_lag.rename(columns={"Moderate Disaster":f"Moderate Disaster Lag {i}",
                                                "Extreme Disaster":f"Extreme Disaster Lag {i}"})

                df_0 = pd.merge(df_0,df_lag,how="outer",on=["Postal Code","Period"])

                moderate_cols.append(f"Moderate Disaster Lag {i}")
                extreme_cols.append(f"Extreme Disaster Lag {i}")

        # Since our dataset only contains months when a disaster happens, missing values means a disaster didn't happen
        df_0 = df_0.fillna(False)

        df_0[f"Moderate Disaster {lag}"] = df_0[moderate_cols].max(axis=1)
        df_0[f"Extreme Disaster {lag}"] = df_0[extreme_cols].max(axis=1)

        # We still want to prevent overlap between extreme and moderate disasters
        df_0.loc[(df_0[f"Moderate Disaster {lag}"]==True) & (df_0[f"Extreme Disaster {lag}"]),f"Moderate Disaster {lag}"] = False

        df_0 = df_0[["Period","Postal Code",f"Moderate Disaster {lag}",f"Extreme Disaster {lag}"]]

        return df_0


data = pd.merge(data,housing_assistance,on=["Period","Postal Code"],how="left")
data["Moderate Disaster"] = data["Moderate Disaster"].fillna(False)
data["Extreme Disaster"] = data["Extreme Disaster"].fillna(False)
data["Uncovered Damage Ratio"] = data["Uncovered Damage Ratio"].fillna(0)

disaster_3 = disaster_lag(housing_assistance.loc[:, housing_assistance.columns != "Uncovered Damage Ratio"],3)
data = pd.merge(data,disaster_3,on=["Period","Postal Code"],how="left")
data["Moderate Disaster 3"] = data["Moderate Disaster 3"].fillna(False)
data["Extreme Disaster 3"] = data["Extreme Disaster 3"].fillna(False)

disaster_6 = disaster_lag(housing_assistance.loc[:, housing_assistance.columns != "Uncovered Damage Ratio"],6)
data = pd.merge(data,disaster_6,on=["Period","Postal Code"],how="left")
data["Moderate Disaster 6"] = data["Moderate Disaster 6"].fillna(False)
data["Extreme Disaster 6"] = data["Extreme Disaster 6"].fillna(False)

disaster_9 = disaster_lag(housing_assistance.loc[:, housing_assistance.columns != "Uncovered Damage Ratio"],9)
data = pd.merge(data,disaster_9,on=["Period","Postal Code"],how="left")
data["Moderate Disaster 9"] = data["Moderate Disaster 9"].fillna(False)
data["Extreme Disaster 9"] = data["Extreme Disaster 9"].fillna(False)




del disasters
del disaster_times
del disaster_3
del disaster_6
del disaster_9
del housing_assistance
```

```python
# Finally, we convert indicator variables to 1 or 0 columns
data["Primary Residence"] = (data["Occupancy Status"] == "P").astype(int)
data.drop(columns=["Occupancy Status"],inplace=True)
data["First Time Homebuyer Indicator"] = (data["First Time Homebuyer Indicator"] == "Y").astype(int)
data["HARP Indicator"] = (data["HARP Indicator"] == "Y").astype(int)
data["Default Next 12"] = data["Default Next 12"].astype(int)
data["Extreme Disaster"] = data["Extreme Disaster"].astype(int)
data["Moderate Disaster"] = data["Moderate Disaster"].astype(int)
data["Extreme Disaster 3"] = data["Extreme Disaster 3"].astype(int)
data["Moderate Disaster 3"] = data["Moderate Disaster 3"].astype(int)
data["Extreme Disaster 6"] = data["Extreme Disaster 6"].astype(int)
data["Moderate Disaster 6"] = data["Moderate Disaster 6"].astype(int)
data["Extreme Disaster 9"] = data["Extreme Disaster 9"].astype(int)
data["Moderate Disaster 9"] = data["Moderate Disaster 9"].astype(int)


con = sqlite3.connect(database="data.sqlite")
(data
  .to_sql(name="cleaned_data", 
          con=con, 
          if_exists="replace",
          index=False)
)
con.close()
```
